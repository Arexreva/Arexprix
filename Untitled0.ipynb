# Laboratory 1 — Deep Neural Networks from Scratch
### Fashion-MNIST · NumPy backprop + PyTorch/Keras experiments

#### Dataset Description
--------------------
* File:        fashion-mnist_test.csv (from fashion-mnist_test_csv.zip)
* Source:      Fashion-MNIST (Zalando Research) — Kaggle CSV format
* Samples:     10,000 grayscale images (this file is the TEST split only;
               the full dataset has 60,000 training + 10,000 test images)
* Image size:  28 x 28 pixels, single channel (grayscale)

* Structure:
    * Column 1       : "label"     -> integer class ID (0-9)
    * Columns 2-785   : "pixel1" ... "pixel784" -> flattened pixel intensities
                      (0-255), row-major order (i.e. pixel_i corresponds to
                      row (i-1)//28, column (i-1)%28 of the 28x28 image)

* Classes (label -> garment):
    * 0 - T-shirt/top     5 - Sandal
    * 1 - Trouser         6 - Shirt
    * 2 - Pullover        7 - Sneaker
    * 3 - Dress           8 - Bag
    * 4 - Coat            9 - Ankle boot

* Notes:
    * This CSV contains only the test split (10,000 rows). Training was
      instead done using tf.keras.datasets.fashion_mnist.load_data(), which
      downloads the full 60,000/10,000 train/test split directly.
    * Pixel values must be normalized (divide by 255.0) before feeding into
      the network, and labels must be one-hot encoded for the softmax output.

## 1.1 NumPy Implementation

```python
import numpy as np


# ---------- Initialization ----------
def init_params(n_x, n_h, n_y, seed=42):
    rng = np.random.default_rng(seed)
    W1 = rng.standard_normal((n_x, n_h)) * np.sqrt(2.0 / n_x)   # He init (good with ReLU)
    b1 = np.zeros((1, n_h))
    W2 = rng.standard_normal((n_h, n_y)) * np.sqrt(2.0 / n_h)
    b2 = np.zeros((1, n_y))
    return {"W1": W1, "b1": b1, "W2": W2, "b2": b2}


# ---------- Forward pass ----------
def relu(Z):
    return np.maximum(0, Z)


def softmax(Z):
    Z = Z - np.max(Z, axis=1, keepdims=True)   # numerical stability
    expZ = np.exp(Z)
    return expZ / np.sum(expZ, axis=1, keepdims=True)


def forward(X, params):
    Z1 = X @ params["W1"] + params["b1"]
    A1 = relu(Z1)
    Z2 = A1 @ params["W2"] + params["b2"]
    A2 = softmax(Z2)
    cache = {"Z1": Z1, "A1": A1, "Z2": Z2, "A2": A2}
    return A2, cache


def compute_loss(A2, Y_onehot):
    m = Y_onehot.shape[0]
    eps = 1e-12
    return -np.sum(Y_onehot * np.log(A2 + eps)) / m


# ---------- Backward pass ----------
def backward(X, Y_onehot, params, cache):
    m = X.shape[0]
    A1, A2 = cache["A1"], cache["A2"]
    Z1 = cache["Z1"]

    dZ2 = (A2 - Y_onehot) / m                 # (m, n_y) — softmax+CE gradient simplifies to this
    dW2 = A1.T @ dZ2                          # (n_h, n_y)
    db2 = np.sum(dZ2, axis=0, keepdims=True)  # (1, n_y)

    dA1 = dZ2 @ params["W2"].T                # (m, n_h)
    dZ1 = dA1 * (Z1 > 0)                      # ReLU derivative
    dW1 = X.T @ dZ1                           # (n_x, n_h)
    db1 = np.sum(dZ1, axis=0, keepdims=True)  # (1, n_h)

    return {"dW1": dW1, "db1": db1, "dW2": dW2, "db2": db2}


# ---------- Numerical gradient check (graded checkpoint) ----------
def gradient_check(X, Y_onehot, params, epsilon=1e-5, n_checks=10, seed=0):
    rng = np.random.default_rng(seed)
    A2, cache = forward(X, params)
    grads = backward(X, Y_onehot, params, cache)

    max_rel_error = 0.0
    for name in ["W1", "b1", "W2", "b2"]:
        param = params[name]
        grad = grads["d" + name]
        flat_idx = rng.choice(param.size, size=min(n_checks, param.size), replace=False)

        for idx in flat_idx:
            multi_idx = np.unravel_index(idx, param.shape)

            orig = param[multi_idx]

            param[multi_idx] = orig + epsilon
            A2_plus, _ = forward(X, params)
            loss_plus = compute_loss(A2_plus, Y_onehot)

            param[multi_idx] = orig - epsilon
            A2_minus, _ = forward(X, params)
            loss_minus = compute_loss(A2_minus, Y_onehot)

            param[multi_idx] = orig  # restore

            numerical_grad = (loss_plus - loss_minus) / (2 * epsilon)
            analytical_grad = grad[multi_idx]

            denom = abs(numerical_grad) + abs(analytical_grad) + 1e-12
            rel_error = abs(numerical_grad - analytical_grad) / denom
            max_rel_error = max(max_rel_error, rel_error)

    return max_rel_error


# ---------- Mini-batch gradient descent training loop ----------
def train(X_train, Y_train_onehot, X_val, Y_val_onehot,
          n_h=128, lr=0.1, epochs=20, batch_size=128, seed=42):
    n_x = X_train.shape[1]
    n_y = Y_train_onehot.shape[1]
    params = init_params(n_x, n_h, n_y, seed=seed)

    history = {"train_loss": [], "val_loss": [], "val_acc": []}
    rng = np.random.default_rng(seed)
    m = X_train.shape[0]

    for epoch in range(epochs):
        perm = rng.permutation(m)
        X_shuf, Y_shuf = X_train[perm], Y_train_onehot[perm]

        for start in range(0, m, batch_size):
            end = start + batch_size
            X_batch = X_shuf[start:end]
            Y_batch = Y_shuf[start:end]

            A2, cache = forward(X_batch, params)
            grads = backward(X_batch, Y_batch, params, cache)

            for p in ["W1", "b1", "W2", "b2"]:
                params[p] -= lr * grads["d" + p]

        # ---- end of epoch: log metrics ----
        A2_train, _ = forward(X_train, params)
        train_loss = compute_loss(A2_train, Y_train_onehot)

        A2_val, _ = forward(X_val, params)
        val_loss = compute_loss(A2_val, Y_val_onehot)
        val_preds = np.argmax(A2_val, axis=1)
        val_true = np.argmax(Y_val_onehot, axis=1)
        val_acc = np.mean(val_preds == val_true)

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["val_acc"].append(val_acc)

        print(f"epoch {epoch+1:2d}/{epochs}  "
              f"train_loss={train_loss:.4f}  val_loss={val_loss:.4f}  val_acc={val_acc:.4f}")

    return params, history
```

## 1.2 Run the gradient check (graded checkpoint)

```python
small_X = X_train_np[:20]
small_Y = Y_train_np[:20]
params_check = init_params(784, 32, 10)
err = gradient_check(small_X, small_Y, params_check)
print(f"max relative error: {err:.2e}")
assert err < 1e-6, "Gradient check failed!"
print("✓ Gradient check passed — backprop is correct.")
```

## 2.1 Loss functions: MSE vs cross-entropy

```python
# Trouser = class 1, Sneaker = class 7 in Fashion-MNIST
def get_binary_subset(X, y, class_a=1, class_b=7):
    mask = (y == class_a) | (y == class_b)
    X_sub = X[mask].reshape(mask.sum(), -1).astype(np.float64) / 255.0
    y_sub = (y[mask] == class_b).astype(np.float64)   # 0 = Trouser, 1 = Sneaker
    return X_sub, y_sub

X_bin_train, y_bin_train = get_binary_subset(X_train, y_train)
X_bin_val, y_bin_val = get_binary_subset(X_val, y_val)

print("Binary train set:", X_bin_train.shape, "| positives (Sneaker):", int(y_bin_train.sum()))
```

## 2.2 Train one logistic neuron with each loss function

```python
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

def train_logistic_neuron(X, y, loss_type="bce", lr=0.5, epochs=50, seed=42):
    rng = np.random.default_rng(seed)
    n_features = X.shape[1]
    w = rng.standard_normal(n_features) * 0.01
    b = 0.0
    m = X.shape[0]
    loss_history = []

    for _ in range(epochs):
        z = X @ w + b
        a = sigmoid(z)

        if loss_type == "bce":
            eps = 1e-12
            loss = -np.mean(y * np.log(a + eps) + (1 - y) * np.log(1 - a + eps))
            dz = (a - y) / m                          # BCE gradient — no sigmoid-derivative term
        elif loss_type == "mse":
            loss = np.mean((a - y) ** 2)
            dz = ((a - y) * a * (1 - a)) / m * 2       # MSE gradient — has the a*(1-a) saturation term
        else:
            raise ValueError("loss_type must be 'bce' or 'mse'")

        dw = X.T @ dz
        db = np.sum(dz)
        w -= lr * dw
        b -= lr * db
        loss_history.append(loss)

    return w, b, loss_history

def accuracy(X, y, w, b):
    preds = (sigmoid(X @ w + b) >= 0.5).astype(np.float64)
    return np.mean(preds == y)

w_bce, b_bce, loss_bce = train_logistic_neuron(X_bin_train, y_bin_train, loss_type="bce", lr=0.5, epochs=50)
w_mse, b_mse, loss_mse = train_logistic_neuron(X_bin_train, y_bin_train, loss_type="mse", lr=0.5, epochs=50)

print(f"Final BCE loss: {loss_bce[-1]:.4f}  | val accuracy: {accuracy(X_bin_val, y_bin_val, w_bce, b_bce):.4f}")
print(f"Final MSE loss: {loss_mse[-1]:.4f}  | val accuracy: {accuracy(X_bin_val, y_bin_val, w_mse, b_mse):.4f}")
```

## Exercise 4 — Hyperparameter analysis


## 4.1 Learning rate

```python
import time
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

SEED = 42


def set_seed():
    np.random.seed(SEED)
    tf.random.set_seed(SEED)


def build_mlp(hidden_units=(128,), activation="relu", init="he_normal",
              optimizer=None, dropout=0.0, l2=0.0, input_dim=784, n_classes=10):
    """Generic MLP builder used by every 4.x sweep."""
    reg = tf.keras.regularizers.l2(l2) if l2 > 0 else None
    layers = [tf.keras.layers.Input(shape=(input_dim,))]
    for units in hidden_units:
        layers.append(tf.keras.layers.Dense(units, activation=activation,
                                             kernel_initializer=init,
                                             kernel_regularizer=reg))
        if dropout > 0:
            layers.append(tf.keras.layers.Dropout(dropout))
    layers.append(tf.keras.layers.Dense(n_classes, activation="softmax"))
    model = tf.keras.Sequential(layers)

    if optimizer is None:
        optimizer = tf.keras.optimizers.Adam(learning_rate=1e-3)
    model.compile(optimizer=optimizer, loss="categorical_crossentropy", metrics=["accuracy"])
    return model


def run_4_1_learning_rate(X_train_k, y_train_k, X_val_k, y_val_k, epochs=20):
    lrs = [1e-4, 1e-3, 1e-2, 1e-1]
    histories = {}
    for lr in lrs:
        set_seed()
        model = build_mlp(optimizer=tf.keras.optimizers.SGD(learning_rate=lr))
        h = model.fit(X_train_k, y_train_k, validation_data=(X_val_k, y_val_k),
                      epochs=epochs, batch_size=128, verbose=0)
        histories[lr] = h.history
        print(f"lr={lr:.0e}  final val_acc={h.history['val_accuracy'][-1]:.4f}")

    plt.figure(figsize=(7, 5))
    for lr, h in histories.items():
        plt.plot(h["val_accuracy"], label=f"lr={lr:.0e}")
    plt.xlabel("epoch"); plt.ylabel("val accuracy"); plt.legend()
    plt.title("4.1 Learning rate sweep")
    plt.savefig("ex4_1_learning_rate.png", dpi=150)
    plt.show()
    return histories


def run_4_2_batch_regimes(X_train, y_train, epochs=15, subset_size=8000):
    X_sub = X_train[:subset_size].reshape(subset_size, -1).astype("float32") / 255.0
    y_sub = tf.keras.utils.to_categorical(y_train[:subset_size], 10)

    regimes = {"stochastic (B=1)": 1, "mini-batch (B=128)": 128, "full-batch (B=N)": subset_size}
    histories, timings, updates_per_epoch = {}, {}, {}

    for name, batch_size in regimes.items():
        set_seed()
        model = build_mlp(optimizer=tf.keras.optimizers.SGD(learning_rate=0.1))
        start = time.time()
        h = model.fit(X_sub, y_sub, epochs=epochs, batch_size=batch_size, verbose=0)
        elapsed = time.time() - start

        histories[name] = h.history
        timings[name] = elapsed
        updates_per_epoch[name] = int(np.ceil(subset_size / batch_size))
        print(f"{name:22s}  updates/epoch={updates_per_epoch[name]:6d}  "
              f"wall_time={elapsed:6.1f}s  final_train_loss={h.history['loss'][-1]:.4f}")

    plt.figure(figsize=(7, 5))
    for name, h in histories.items():
        plt.plot(h["loss"], label=name)
    plt.xlabel("epoch"); plt.ylabel("training loss"); plt.legend()
    plt.title("4.2 Training loss by batching regime")
    plt.savefig("ex4_2_batch_loss.png", dpi=150)
    plt.show()

    plt.figure(figsize=(7, 5))
    for name, h in histories.items():
        plt.plot(h["accuracy"], label=name)
    plt.xlabel("epoch"); plt.ylabel("train accuracy"); plt.legend()
    plt.title("4.2 Train accuracy by batching regime")
    plt.savefig("ex4_2_batch_acc.png", dpi=150)
    plt.show()

    return histories, timings, updates_per_epoch


def run_4_3_depth(X_train_k, y_train_k, X_val_k, y_val_k, epochs=20):
    depths = [1, 2, 3, 5]
    histories = {}
    for d in depths:
        set_seed()
        model = build_mlp(hidden_units=tuple([128] * d))
        h = model.fit(X_train_k, y_train_k, validation_data=(X_val_k, y_val_k),
                      epochs=epochs, batch_size=128, verbose=0)
        histories[d] = h.history
        gap = h.history["accuracy"][-1] - h.history["val_accuracy"][-1]
        print(f"depth={d}  final val_acc={h.history['val_accuracy'][-1]:.4f}  train/val gap={gap:.4f}")

    plt.figure(figsize=(7, 5))
    for d, h in histories.items():
        plt.plot(h["val_accuracy"], label=f"{d} hidden layer(s)")
    plt.xlabel("epoch"); plt.ylabel("val accuracy"); plt.legend()
    plt.title("4.3 Depth sweep")
    plt.savefig("ex4_3_depth.png", dpi=150)
    plt.show()
    return histories


def run_4_4_width(X_train_k, y_train_k, X_val_k, y_val_k, epochs=20):
    widths = [32, 128, 512]
    histories = {}
    for w in widths:
        set_seed()
        model = build_mlp(hidden_units=(w,))
        n_params = model.count_params()
        h = model.fit(X_train_k, y_train_k, validation_data=(X_val_k, y_val_k),
                      epochs=epochs, batch_size=128, verbose=0)
        histories[w] = h.history
        gap = h.history["accuracy"][-1] - h.history["val_accuracy"][-1]
        print(f"width={w:4d}  params={n_params:,}  final val_acc={h.history['val_accuracy'][-1]:.4f}  gap={gap:.4f}")

    plt.figure(figsize=(7, 5))
    for w, h in histories.items():
        plt.plot(h["val_accuracy"], label=f"{w} units")
    plt.xlabel("epoch"); plt.ylabel("val accuracy"); plt.legend()
    plt.title("4.4 Width sweep")
    plt.savefig("ex4_4_width.png", dpi=150)
    plt.show()
    return histories


def run_4_5_activation(X_train_k, y_train_k, X_val_k, y_val_k, epochs=20):
    configs = {
        "sigmoid": ("sigmoid", "glorot_normal"),
        "tanh": ("tanh", "glorot_normal"),
        "relu": ("relu", "he_normal"),
        "leaky_relu": (tf.keras.layers.LeakyReLU(negative_slope=0.01), "he_normal"),
    }
    histories = {}
    for name, (act, init) in configs.items():
        set_seed()
        if isinstance(act, str):
            model = build_mlp(activation=act, init=init)
        else:
            model = tf.keras.Sequential([
                tf.keras.layers.Input(shape=(784,)),
                tf.keras.layers.Dense(128, kernel_initializer=init),
                act,
                tf.keras.layers.Dense(10, activation="softmax"),
            ])
            model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
                          loss="categorical_crossentropy", metrics=["accuracy"])
        h = model.fit(X_train_k, y_train_k, validation_data=(X_val_k, y_val_k),
                      epochs=epochs, batch_size=128, verbose=0)
        histories[name] = h.history
        print(f"{name:12s} final val_acc={h.history['val_accuracy'][-1]:.4f}")

    plt.figure(figsize=(7, 5))
    for name, h in histories.items():
        plt.plot(h["val_accuracy"], label=name)
    plt.xlabel("epoch"); plt.ylabel("val accuracy"); plt.legend()
    plt.title("4.5 Activation function sweep")
    plt.savefig("ex4_5_activation.png", dpi=150)
    plt.show()
    return histories


def run_4_6_init(X_train_k, y_train_k, X_val_k, y_val_k, epochs=20):
    inits = {
        "small_constant": tf.keras.initializers.Constant(0.01),
        "large": tf.keras.initializers.RandomNormal(mean=0.0, stddev=1.0),
        "xavier": "glorot_normal",
        "he": "he_normal",
    }
    histories = {}
    for name, init in inits.items():
        set_seed()
        model = build_mlp(hidden_units=(128, 128, 128, 128), activation="sigmoid", init=init)
        h = model.fit(X_train_k, y_train_k, validation_data=(X_val_k, y_val_k),
                      epochs=epochs, batch_size=128, verbose=0)
        histories[name] = h.history
        print(f"{name:15s} final val_acc={h.history['val_accuracy'][-1]:.4f}")

    plt.figure(figsize=(7, 5))
    for name, h in histories.items():
        plt.plot(h["val_accuracy"], label=name)
    plt.xlabel("epoch"); plt.ylabel("val accuracy"); plt.legend()
    plt.title("4.6 Weight initialization sweep (4-layer sigmoid net)")
    plt.savefig("ex4_6_init.png", dpi=150)
    plt.show()
    return histories


def run_4_7_optimizer(X_train_k, y_train_k, X_val_k, y_val_k, epochs=20):
    optimizers = {
        "SGD": tf.keras.optimizers.SGD(learning_rate=1e-3),
        "SGD+momentum": tf.keras.optimizers.SGD(learning_rate=1e-3, momentum=0.9),
        "RMSProp": tf.keras.optimizers.RMSprop(learning_rate=1e-3),
        "Adam": tf.keras.optimizers.Adam(learning_rate=1e-3),
    }
    histories = {}
    for name, opt in optimizers.items():
        set_seed()
        model = build_mlp(optimizer=opt)
        h = model.fit(X_train_k, y_train_k, validation_data=(X_val_k, y_val_k),
                      epochs=epochs, batch_size=128, verbose=0)
        histories[name] = h.history
        print(f"{name:14s} final val_acc={h.history['val_accuracy'][-1]:.4f}")

    plt.figure(figsize=(7, 5))
    for name, h in histories.items():
        plt.plot(h["val_accuracy"], label=name)
    plt.xlabel("epoch"); plt.ylabel("val accuracy"); plt.legend()
    plt.title("4.7 Optimizer sweep")
    plt.savefig("ex4_7_optimizer.png", dpi=150)
    plt.show()
    return histories


def run_4_8_regularization(X_train_k, y_train_k, X_val_k, y_val_k, epochs=25):
    configs = {
        "none": dict(dropout=0.0, l2=0.0),
        "L2": dict(dropout=0.0, l2=1e-3),
        "dropout": dict(dropout=0.5, l2=0.0),
        "L2+dropout": dict(dropout=0.5, l2=1e-3),
    }
    histories = {}
    for name, cfg in configs.items():
        set_seed()
        model = build_mlp(hidden_units=(512, 512), **cfg)
        h = model.fit(X_train_k, y_train_k, validation_data=(X_val_k, y_val_k),
                      epochs=epochs, batch_size=128, verbose=0)
        histories[name] = h.history
        best_epoch = int(np.argmin(h.history["val_loss"])) + 1
        print(f"{name:12s} final val_acc={h.history['val_accuracy'][-1]:.4f}  "
              f"min val_loss epoch={best_epoch}")

    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    for name, h in histories.items():
        axes[0].plot(h["val_loss"], label=name)
        axes[1].plot(h["val_accuracy"], label=name)
    axes[0].set_title("4.8 Validation loss"); axes[0].set_xlabel("epoch"); axes[0].legend()
    axes[1].set_title("4.8 Validation accuracy"); axes[1].set_xlabel("epoch"); axes[1].legend()
    plt.savefig("ex4_8_regularization.png", dpi=150)
    plt.show()
    return histories
```
