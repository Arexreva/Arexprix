```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# reproducibility
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

CLASS_NAMES = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

(X_train_full, y_train_full), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

print("X_train_full:", X_train_full.shape, "y_train_full:", y_train_full.shape)
print("X_test:", X_test.shape, "y_test:", y_test.shape)

# carve out a validation set from training data (last 10,000 images)
X_train, y_train = X_train_full[:-10000], y_train_full[:-10000]
X_val, y_val = X_train_full[-10000:], y_train_full[-10000:]

print("X_train:", X_train.shape, "X_val:", X_val.shape, "X_test:", X_test.shape)
```

```python
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for i, ax in enumerate(axes.flat):
    ax.imshow(X_train[i], cmap='gray')
    ax.set_title(CLASS_NAMES[y_train[i]])
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
def preprocess_for_numpy(X, y, n_classes=10):
    X_flat = X.reshape(X.shape[0], -1).astype(np.float64) / 255.0   # normalize to [0,1]
    Y_onehot = np.eye(n_classes)[y]
    return X_flat, Y_onehot

X_train_np, Y_train_np = preprocess_for_numpy(X_train, y_train)
X_val_np, Y_val_np = preprocess_for_numpy(X_val, y_val)
X_test_np, Y_test_np = preprocess_for_numpy(X_test, y_test)

print(X_train_np.shape, Y_train_np.shape)   # (50000, 784) (50000, 10)
```

```python
import numpy as np


# ---------- Initialization ----------
def init_params(n_x, n_h, n_y, seed=42):
    rng = np.random.default_rng(seed)
    W1 = rng.standard_normal((n_x, n_h)) * np.sqrt(2.0 / n_x)   # He init (good with ReLU)
    b1 = np.zeros((1, n_h))
    W2 = rng.standard_normal((n_h, n_y)) * np.sqrt(2.0 / n_h)
    b2 = np.zeros((1, n_y))
    return {"W1": W1, "b1": b1, "W2": W2, "b2": b2}


# ---------- Forward pass ----------
def relu(Z):
    return np.maximum(0, Z)


def softmax(Z):
    Z = Z - np.max(Z, axis=1, keepdims=True)   # numerical stability
    expZ = np.exp(Z)
    return expZ / np.sum(expZ, axis=1, keepdims=True)


def forward(X, params):
    Z1 = X @ params["W1"] + params["b1"]
    A1 = relu(Z1)
    Z2 = A1 @ params["W2"] + params["b2"]
    A2 = softmax(Z2)
    cache = {"Z1": Z1, "A1": A1, "Z2": Z2, "A2": A2}
    return A2, cache


def compute_loss(A2, Y_onehot):
    m = Y_onehot.shape[0]
    eps = 1e-12
    return -np.sum(Y_onehot * np.log(A2 + eps)) / m


# ---------- Backward pass ----------
def backward(X, Y_onehot, params, cache):
    m = X.shape[0]
    A1, A2 = cache["A1"], cache["A2"]
    Z1 = cache["Z1"]

    dZ2 = (A2 - Y_onehot) / m                 # (m, n_y) — softmax+CE gradient simplifies to this
    dW2 = A1.T @ dZ2                          # (n_h, n_y)
    db2 = np.sum(dZ2, axis=0, keepdims=True)  # (1, n_y)

    dA1 = dZ2 @ params["W2"].T                # (m, n_h)
    dZ1 = dA1 * (Z1 > 0)                      # ReLU derivative
    dW1 = X.T @ dZ1                           # (n_x, n_h)
    db1 = np.sum(dZ1, axis=0, keepdims=True)  # (1, n_h)

    return {"dW1": dW1, "db1": db1, "dW2": dW2, "db2": db2}


# ---------- Numerical gradient check (graded checkpoint) ----------
def gradient_check(X, Y_onehot, params, epsilon=1e-5, n_checks=10, seed=0):
    rng = np.random.default_rng(seed)
    A2, cache = forward(X, params)
    grads = backward(X, Y_onehot, params, cache)

    max_rel_error = 0.0
    for name in ["W1", "b1", "W2", "b2"]:
        param = params[name]
        grad = grads["d" + name]
        flat_idx = rng.choice(param.size, size=min(n_checks, param.size), replace=False)

        for idx in flat_idx:
            multi_idx = np.unravel_index(idx, param.shape)

            orig = param[multi_idx]

            param[multi_idx] = orig + epsilon
            A2_plus, _ = forward(X, params)
            loss_plus = compute_loss(A2_plus, Y_onehot)

            param[multi_idx] = orig - epsilon
            A2_minus, _ = forward(X, params)
            loss_minus = compute_loss(A2_minus, Y_onehot)

            param[multi_idx] = orig  # restore

            numerical_grad = (loss_plus - loss_minus) / (2 * epsilon)
            analytical_grad = grad[multi_idx]

            denom = abs(numerical_grad) + abs(analytical_grad) + 1e-12
            rel_error = abs(numerical_grad - analytical_grad) / denom
            max_rel_error = max(max_rel_error, rel_error)

    return max_rel_error


# ---------- Mini-batch gradient descent training loop ----------
def train(X_train, Y_train_onehot, X_val, Y_val_onehot,
          n_h=128, lr=0.1, epochs=20, batch_size=128, seed=42):
    n_x = X_train.shape[1]
    n_y = Y_train_onehot.shape[1]
    params = init_params(n_x, n_h, n_y, seed=seed)

    history = {"train_loss": [], "val_loss": [], "val_acc": []}
    rng = np.random.default_rng(seed)
    m = X_train.shape[0]

    for epoch in range(epochs):
        perm = rng.permutation(m)
        X_shuf, Y_shuf = X_train[perm], Y_train_onehot[perm]

        for start in range(0, m, batch_size):
            end = start + batch_size
            X_batch = X_shuf[start:end]
            Y_batch = Y_shuf[start:end]

            A2, cache = forward(X_batch, params)
            grads = backward(X_batch, Y_batch, params, cache)

            for p in ["W1", "b1", "W2", "b2"]:
                params[p] -= lr * grads["d" + p]

        # ---- end of epoch: log metrics ----
        A2_train, _ = forward(X_train, params)
        train_loss = compute_loss(A2_train, Y_train_onehot)

        A2_val, _ = forward(X_val, params)
        val_loss = compute_loss(A2_val, Y_val_onehot)
        val_preds = np.argmax(A2_val, axis=1)
        val_true = np.argmax(Y_val_onehot, axis=1)
        val_acc = np.mean(val_preds == val_true)

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["val_acc"].append(val_acc)

        print(f"epoch {epoch+1:2d}/{epochs}  "
              f"train_loss={train_loss:.4f}  val_loss={val_loss:.4f}  val_acc={val_acc:.4f}")

    return params, history
```

```python
small_X = X_train_np[:20]
small_Y = Y_train_np[:20]
params_check = init_params(784, 32, 10)
err = gradient_check(small_X, small_Y, params_check)
print(f"max relative error: {err:.2e}")
assert err < 1e-6, "Gradient check failed!"
print("✓ Gradient check passed — backprop is correct.")
```

```python
params, history = train(X_train_np, Y_train_np, X_val_np, Y_val_np,
                         n_h=128, lr=0.1, epochs=20, batch_size=128, seed=SEED)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(history["train_loss"], label="train loss")
axes[0].plot(history["val_loss"], label="val loss")
axes[0].set_xlabel("epoch"); axes[0].set_ylabel("loss"); axes[0].legend(); axes[0].set_title("Loss")

axes[1].plot(history["val_acc"], label="val accuracy", color="green")
axes[1].set_xlabel("epoch"); axes[1].set_ylabel("accuracy"); axes[1].legend(); axes[1].set_title("Validation Accuracy")

plt.tight_layout()
plt.savefig("exercise1_curves.png", dpi=150)
plt.show()

print(f"Final validation accuracy: {history['val_acc'][-1]:.4f}")
```

My gradient check reported a maximum relative error of [paste your printed value, e.g. 1.6e-08]. This is evidence that my backpropagation is correct because the numerical gradient — computed by directly perturbing each parameter by ±ε and measuring the resulting change in loss — approximates the true derivative independently of any analytical formula. If my hand-derived backward() function had a bug (wrong transpose, missing chain-rule term, wrong sign), the analytical gradient would diverge from this numerical estimate and the relative error would be much larger (often close to 1, or even 100% wrong in direction). A relative error below 1e-6 means the two independently-computed gradients agree to about 6 significant digits, which is only possible if the analytical derivation and its implementation are correct — coincidental agreement at that precision is essentially impossible.

Q1.3 — Why does this hand-written network end up less accurate than the framework model you build in Exercise 3?

This is conceptual — you can answer it now, before even running Exercise 3:
My NumPy network uses a single fixed hyperparameter configuration with plain SGD, no momentum, and no adaptive learning rates, whereas the framework model in Exercise 3 uses the Adam optimizer. Adam adapts the effective learning rate per-parameter using running estimates of the first and second moments of the gradients, which converges faster and more reliably than plain SGD, especially early in training. Frameworks also apply numerically stable, well-tested implementations (e.g. fused matrix ops, better numerical precision handling in softmax/cross-entropy) and default good practices (like proper weight initialization scaling) that reduce the chance of subtle inefficiencies. The architecture is identical, so the accuracy gap comes from the optimizer and training dynamics, not from any difference in model capacity.
